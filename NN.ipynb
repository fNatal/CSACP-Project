```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt

print("8")
```

```python
# Load the dataset
df_updated_imputed = pd.read_csv('Updated_Imputed_Dataset.csv')
```

```python




# Select relevant columns: target variables (offender characteristics) and features (victim details)
target_columns = [
    'Perpetrator Sex', 'Perpetrator Age', 'Perpetrator Race', 'Perpetrator Ethnicity', 'Relationship',
    'Victim Count', 'Offender Name', 'Offender Appearance', 'Offender Job', 'Offender Age', 'Offender Height',
    'Offender Weight', 'Offender Personality', 'Offender Education and Family Background', 'Offender Motive',
    'Offender Planned or Unplanned', 'Offender Regular or One-Time', 'Offender No. of Homicides',
    'Offender Part of Organization', 'Victim If Randomly Chosen or Planned']
input_columns = df_updated_imputed.columns.difference(target_columns)  # All columns except the target columns

# Separate features (X) and target (y)
X = df_updated_imputed[input_columns]  # Features: victim details and others
y = df_updated_imputed[target_columns]  # Target: offender characteristics

# Handle categorical columns by encoding them (if any)
categorical_columns = X.select_dtypes(include=['object']).columns
for col in categorical_columns:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))  # Convert categorical variables to numeric

# Handle the categorical columns in the target variable
for col in target_columns:
    if y[col].dtype == 'object':  # Check if it's categorical
        le = LabelEncoder()
        y[col] = le.fit_transform(y[col].astype(str))

# Normalize the data (important for Neural Networks)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split the data into training and testing sets (80% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Build the Neural Network model with 6 hidden layers
model = Sequential()

# Input layer (same shape as the number of features)
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))  # 128 neurons in the input layer

# Hidden layers (6 hidden layers)
model.add(Dense(256, activation='relu'))  # First hidden layer
model.add(Dense(128, activation='relu'))  # Second hidden layer
model.add(Dense(64, activation='relu'))   # Third hidden layer
model.add(Dense(32, activation='relu'))   # Fourth hidden layer
model.add(Dense(16, activation='relu'))   # Fifth hidden layer
model.add(Dense(8, activation='relu'))    # Sixth hidden layer

# Output layer: predicting multiple continuous values (offender characteristics)
model.add(Dense(y_train.shape[1]))  # Output neurons same as number of target variables

# Compile the model
model.compile(optimizer=Adam(), loss='mean_squared_error', metrics=['accuracy'])

# Train the model
history = model.fit(X_train, y_train, epochs=100, batch_size=32, validation_split=0.2)

# Evaluate the model on the test set
test_loss, test_accuracy = model.evaluate(X_test, y_test)
print(f'Test Loss: {test_loss}')
print(f'Test Accuracy: {test_accuracy}')

# Predict the offender characteristics using the trained model
predictions = model.predict(X_test)

# Display the predictions (first 5 predictions)
print("Predictions (first 5):")
print(predictions[:5])

# Optionally, you can save the trained model
# model.save('offender_characteristics_model.h5')
```
